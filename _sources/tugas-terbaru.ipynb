# A. Soal Determinan Matriks 3×3
## Soal 1
### Langkah Penyelesaian (Metode Sarrus)

Untuk menghitung determinan matriks $3\times3$, kita dapat menggunakan **Metode Sarrus**. Caranya adalah dengan menuliskan kembali dua kolom pertama di sebelah kanan matriks asli:

$$\begin{matrix}
2 & 1 & 3 & | & 2 & 1 \\
0 & 4 & 5 & | & 0 & 4 \\
1 & 2 & 1 & | & 1 & 2
\end{matrix}$$

#### 1. Jumlahkan Hasil Kali Diagonal Utama (\\)
* Diagonal 1: $2 \times 4 \times 1 = 8$
* Diagonal 2: $1 \times 5 \times 1 = 5$
* Diagonal 3: $3 \times 0 \times 2 = 0$

$$\text{Total Diagonal Utama} = 8 + 5 + 0 = 13$$

#### 2. Jumlahkan Hasil Kali Diagonal Sekunder (/)
* Diagonal 1: $1 \times 4 \times 3 = 12$
* Diagonal 2: $2 \times 5 \times 2 = 20$
* Diagonal 3: $1 \times 0 \times 1 = 0$

$$\text{Total Diagonal Sekunder} = 12 + 20 + 0 = 32$$

#### 3. Hitung Determinan
$$\det(A) = \text{Total Diagonal Utama} - \text{Total Diagonal Sekunder}$$
$$\det(A) = 13 - 32$$
$$\det(A) = -19$$

---

**Kesimpulan:**
Nilai determinan dari matriks $A$ adalah **$-19$**.

## Soal 2
### Langkah Penyelesaian (Metode Sarrus)

Untuk menghitung determinan matriks $3\times3$ dari matriks $B$, kita dapat menuliskan kembali dua kolom pertama di sebelah kanan matriks tersebut:

$$\begin{matrix}
3 & 2 & 1 & | & 3 & 2 \\
1 & 0 & 4 & | & 1 & 0 \\
2 & 5 & 1 & | & 2 & 5
\end{matrix}$$

#### 1. Jumlahkan Hasil Kali Diagonal Utama (dari kiri atas ke kanan bawah)
* Diagonal 1: $3 \times 0 \times 1 = 0$
* Diagonal 2: $2 \times 4 \times 2 = 16$
* Diagonal 3: $1 \times 1 \times 5 = 5$

$$\text{Total Diagonal Utama} = 0 + 16 + 5 = 21$$

#### 2. Jumlahkan Hasil Kali Diagonal Sekunder (dari kiri bawah ke kanan atas)
* Diagonal 1: $2 \times 0 \times 1 = 0$
* Diagonal 2: $5 \times 4 \times 3 = 60$
* Diagonal 3: $1 \times 1 \times 2 = 2$

$$\text{Total Diagonal Sekunder} = 0 + 60 + 2 = 62$$

#### 3. Hitung Nilai Determinan Matriks B
$$\det(B) = \text{Total Diagonal Utama} - \text{Total Diagonal Sekunder}$$
$$\det(B) = 21 - 62$$
$$\det(B) = -41$$

---

**Kesimpulan:**
Nilai determinan dari matriks $B$ adalah **$-41$**.

## Soal 3
### Langkah Penyelesaian

Diketahui matriks:
$$C = \begin{bmatrix} 1 & 2 & 3 \\ 2 & 4 & 6 \\ 1 & 1 & 1 \end{bmatrix}$$

#### 1. Menghitung Nilai $\det(C)$ menggunakan Metode Sarrus
Tuliskan kembali dua kolom pertama di sebelah kanan matriks:

$$\begin{matrix}
1 & 2 & 3 & | & 1 & 2 \\
2 & 4 & 6 & | & 2 & 4 \\
1 & 1 & 1 & | & 1 & 1
\end{matrix}$$

* **Jumlah Diagonal Utama (\\):**
  $$\text{Diagonal 1} = 1 \times 4 \times 1 = 4$$
  $$\text{Diagonal 2} = 2 \times 6 \times 1 = 12$$
  $$\text{Diagonal 3} = 3 \times 2 \times 1 = 6$$
  $$\text{Total Diagonal Utama} = 4 + 12 + 6 = 22$$

* **Jumlah Diagonal Sekunder (/):**
  $$\text{Diagonal 1} = 1 \times 4 \times 3 = 12$$
  $$\text{Diagonal 2} = 1 \times 6 \times 1 = 6$$
  $$\text{Diagonal 3} = 1 \times 2 \times 2 = 4$$
  $$\text{Total Diagonal Sekunder} = 12 + 6 + 4 = 22$$

* **Hitung Determinan:**
  $$\det(C) = \text{Total Diagonal Utama} - \text{Total Diagonal Sekunder}$$
  $$\det(C) = 22 - 22$$
  $$\det(C) = 0$$

> **Cara Cepat Alternatif (Sifat Matriks):** > Perhatikan baris kedua ($R_2$) merupakan kelipatan dari baris pertama ($R_1$), yaitu $R_2 = 2 \times R_1$. Jika suatu matriks memiliki baris atau kolom yang saling berkelipatan (bergantung linier), maka determinannya **pasti 0**.

---

#### 2. Penjelasan Apakah Matriks Termasuk Singular atau Tidak

Berdasarkan definisi dalam aljabar linear:
* **Matriks Singular** adalah matriks yang nilai determinannya sama dengan nol ($\det = 0$). Matriks jenis ini tidak memiliki invers.
* **Matriks Non-Singular** adalah matriks yang nilai determinannya tidak sama dengan nol ($\det \neq 0$).

Karena hasil perhitungan di atas menunjukkan bahwa $\det(C) = 0$, maka matriks $C$ adalah **Matriks Singular**.

---

**Kesimpulan:**
* $\det(C) = 0$
* Matriks tersebut merupakan **matriks singular** karena nilai determinannya bernilai nol.

# B. Soal Dekomposisi Matriks (LU Decomposition)
## Soal 4
### Langkah Penyelesaian (Dekomposisi LU)

Diketahui matriks:
$$A = \begin{bmatrix} 2 & 4 & 2 \\ 1 & 5 & 2 \\ 1 & 2 & 4 \end{bmatrix}$$

Kita ingin memecah matriks $A$ menjadi $A = LU$, di mana:
* $L$ (Lower Triangular) memiliki komponen utama bernilai 1 pada diagonalnya:
  $$L = \begin{bmatrix} 1 & 0 & 0 \\ l_{21} & 1 & 0 \\ l_{31} & l_{32} & 1 \end{bmatrix}$$
* $U$ (Upper Triangular) berbentuk:
  $$U = \begin{bmatrix} u_{11} & u_{12} & u_{13} \\ 0 & u_{22} & u_{23} \\ 0 & 0 & u_{33} \end{bmatrix}$$

Kita akan mencari elemen-elemen tersebut dengan melakukan operasi baris eliminasi Gauss pada matriks $A$ untuk mengubahnya menjadi matriks segitiga atas ($U$), sekaligus mencatat pengali (*multiplier*) untuk mengisi matriks $L$.

---

#### Langkah 1: Mengeliminasi Baris 2 dan Baris 3 pada Kolom 1

* **Untuk Baris 2 ($R_2$):**
  Pengali agar elemen pertama $R_2$ menjadi 0 adalah:
  $$l_{21} = \frac{a_{21}}{a_{11}} = \frac{1}{2} = 0.5$$
  Operasi baris baru: $R_2 \leftarrow R_2 - 0.5 \times R_1$
  $$R_2 = [1, 5, 2] - 0.5 \times [2, 4, 2] = [0, 3, 1]$$

* **Untuk Baris 3 ($R_3$):**
  Pengali agar elemen pertama $R_3$ menjadi 0 adalah:
  $$l_{31} = \frac{a_{31}}{a_{11}} = \frac{1}{2} = 0.5$$
  Operasi baris baru: $R_3 \leftarrow R_3 - 0.5 \times R_1$
  $$R_3 = [1, 2, 4] - 0.5 \times [2, 4, 2] = [0, 0, 3]$$

Setelah Langkah 1, matriks kita bertransformasi menjadi:
$$\begin{bmatrix} 2 & 4 & 2 \\ 0 & 3 & 1 \\ 0 & 0 & 3 \end{bmatrix}$$

---

#### Langkah 2: Mengeliminasi Baris 3 pada Kolom 2

Kebetulan, dari operasi di Langkah 1, elemen pada posisi $(3,2)$ sudah bernilai $0$. 
Maka kita tidak perlu melakukan eliminasi lebih lanjut untuk kolom kedua.
* Pengali untuk posisi ini adalah:
  $$l_{32} = \frac{0}{3} = 0$$

Matriks hasil eliminasi akhir ini langsung menjadi matriks segitiga atas **$U$**:
$$U = \begin{bmatrix} 2 & 4 & 2 \\ 0 & 3 & 1 \\ 0 & 0 & 3 \end{bmatrix}$$

---

#### Langkah 3: Menyusun Matriks $L$

Masukkan nilai pengali $l_{21} = 0.5$, $l_{31} = 0.5$, dan $l_{32} = 0$ ke dalam format matriks $L$:
$$L = \begin{bmatrix} 1 & 0 & 0 \\ 0.5 & 1 & 0 \\ 0.5 & 0 & 1 \end{bmatrix}$$

---

### Pembuktian ($A = LU$)

Mari kita uji kembali hasil perkalian matriks $L$ dan $U$:
$$LU = \begin{bmatrix} 1 & 0 & 0 \\ 0.5 & 1 & 0 \\ 0.5 & 0 & 1 \end{bmatrix} \begin{bmatrix} 2 & 4 & 2 \\ 0 & 3 & 1 \\ 0 & 0 & 3 \end{bmatrix}$$

* Baris 1:
  * $(1\times2) + (0\times0) + (0\times0) = 2$
  * $(1\times4) + (0\times3) + (0\times0) = 4$
  * $(1\times2) + (0\times1) + (0\times0) = 2$
* Baris 2:
  * $(0.5\times2) + (1\times0) + (0\times0) = 1$
  * $(0.5\times4) + (1\times3) + (0\times0) = 2 + 3 = 5$
  * $(0.5\times2) + (1\times1) + (0\times0) = 1 + 1 = 2$
* Baris 3:
  * $(0.5\times2) + (0\times0) + (1\times0) = 1$
  * $(0.5\times4) + (0\times3) + (1\times0) = 2$
  * $(0.5\times2) + (0\times1) + (1\times3) = 1 + 3 = 4$

Hasil perkalian menghasilkan matriks $\begin{bmatrix} 2 & 4 & 2 \\ 1 & 5 & 2 \\ 1 & 2 & 4 \end{bmatrix}$, yang sama persis dengan matriks $A$ semula.

---

### Kesimpulan

Hasil dekomposisi $LU$ dari matriks $A$ adalah:
$$L = \begin{bmatrix} 1 & 0 & 0 \\ 0.5 & 1 & 0 \\ 0.5 & 0 & 1 \end{bmatrix}, \quad U = \begin{bmatrix} 2 & 4 & 2 \\ 0 & 3 & 1 \\ 0 & 0 & 3 \end{bmatrix}$$

## Soal 5
### Langkah Penyelesaian (Dekomposisi LU via Eliminasi Gauss)

Diketahui matriks:
$$B = \begin{bmatrix} 1 & 2 & 1 \\ 2 & 5 & 3 \\ 4 & 10 & 8 \end{bmatrix}$$

Kita akan mentransformasikan matriks $B$ menjadi matriks segitiga atas ($U$) melalui eliminasi Gauss. Nilai pengali (*multiplier*) yang digunakan untuk membuat elemen di bawah diagonal menjadi nol akan kita simpan untuk menyusun matriks segitiga bawah ($L$).

Format dasar matriks $L$ dengan diagonal utama bernilai 1 adalah:
$$L = \begin{bmatrix} 1 & 0 & 0 \\ l_{21} & 1 & 0 \\ l_{31} & l_{32} & 1 \end{bmatrix}$$

---

#### Langkah 1: Eliminasi Kolom 1 (Membuat nol elemen di bawah $b_{11}$)

* **Untuk Baris 2 ($R_2$):**
  Pengali untuk mengeliminasi elemen pertama di baris 2:
  $$l_{21} = \frac{b_{21}}{b_{11}} = \frac{2}{1} = 2$$
  Operasi baris: $R_2 \leftarrow R_2 - 2 \times R_1$
  $$R_2 = [2, 5, 3] - 2 \times [1, 2, 1] = [0, 1, 1]$$

* **Untuk Baris 3 ($R_3$):**
  Pengali untuk mengeliminasi elemen pertama di baris 3:
  $$l_{31} = \frac{b_{31}}{b_{11}} = \frac{4}{1} = 4$$
  Operasi baris: $R_3 \leftarrow R_3 - 4 \times R_1$
  $$R_3 = [4, 10, 8] - 4 \times [1, 2, 1] = [0, 2, 4]$$

Matriks setelah Langkah 1 menjadi:
$$\begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 0 & 2 & 4 \end{bmatrix}$$

---

#### Langkah 2: Eliminasi Kolom 2 (Membuat nol elemen di bawah posisi $(2,2)$)

Gunakan baris kedua yang baru untuk mengeliminasi elemen kedua pada baris ketiga.
* **Untuk Baris 3 ($R_3$):**
  Pengali untuk mengeliminasi elemen kedua di baris 3:
  $$l_{32} = \frac{2}{1} = 2$$
  Operasi baris: $R_3 \leftarrow R_3 - 2 \times R_2$
  $$R_3 = [0, 2, 4] - 2 \times [0, 1, 1] = [0, 0, 2]$$

Matriks hasil akhir ini merupakan matriks segitiga atas **$U$**:
$$U = \begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 0 & 0 & 2 \end{bmatrix}$$

---

#### Langkah 3: Menyusun Matriks $L$

Masukkan nilai pengali yang sudah kita dapatkan ($l_{21} = 2$, $l_{31} = 4$, dan $l_{32} = 2$) ke dalam matriks $L$:
$$L = \begin{bmatrix} 1 & 0 & 0 \\ 2 & 1 & 0 \\ 4 & 2 & 1 \end{bmatrix}$$

---

### Pembuktian ($B = LU$)

Mari kita kalikan kembali matriks $L$ dan $U$ untuk memastikan hasilnya sama dengan matriks $B$:
$$LU = \begin{bmatrix} 1 & 0 & 0 \\ 2 & 1 & 0 \\ 4 & 2 & 1 \end{bmatrix} \begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 0 & 0 & 2 \end{bmatrix}$$

* Baris 1:
  * $(1\times1) + (0\times0) + (0\times0) = 1$
  * $(1\times2) + (0\times1) + (0\times0) = 2$
  * $(1\times1) + (0\times1) + (0\times0) = 1$
* Baris 2:
  * $(2\times1) + (1\times0) + (0\times0) = 2$
  * $(2\times2) + (1\times1) + (0\times0) = 4 + 1 = 5$
  * $(2\times1) + (1\times1) + (0\times0) = 2 + 1 = 3$
* Baris 3:
  * $(4\times1) + (2\times0) + (1\times0) = 4$
  * $(4\times2) + (2\times1) + (1\times0) = 8 + 2 = 10$
  * $(4\times1) + (2\times1) + (1\times2) = 4 + 2 + 2 = 8$

Hasil perkalian menghasilkan matriks $\begin{bmatrix} 1 & 2 & 1 \\ 2 & 5 & 3 \\ 4 & 10 & 8 \end{bmatrix}$, cocok dengan matriks $B$ mula-mula.

---

### Kesimpulan

Matriks $L$ dan $U$ dari matriks $B$ adalah:
$$L = \begin{bmatrix} 1 & 0 & 0 \\ 2 & 1 & 0 \\ 4 & 2 & 1 \end{bmatrix}, \quad U = \begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 0 & 0 & 2 \end{bmatrix}$$

# C. Soal Invers Matriks 3×3
## Soal 7
### Langkah Penyelesaian (Mencari Invers dengan Metode Adjoin)

Diketahui matriks:
$$A = \begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 2 & 3 & 4 \end{bmatrix}$$

Rumus invers matriks adalah:
$$A^{-1} = \frac{1}{\det(A)} \cdot \text{Adj}(A)$$

---

#### Langkah 1: Menghitung Determinan Matriks $A$ ($\det(A)$)
Menggunakan Metode Sarrus:
$$\begin{matrix}
1 & 2 & 1 & | & 1 & 2 \\
0 & 1 & 1 & | & 0 & 1 \\
2 & 3 & 4 & | & 2 & 3
\end{matrix}$$

* **Total Diagonal Utama (\\):**
  $$(1 \times 1 \times 4) + (2 \times 1 \times 2) + (1 \times 0 \times 3) = 4 + 4 + 0 = 8$$
* **Total Diagonal Sekunder (/):**
  $$(2 \times 1 \times 1) + (3 \times 1 \times 1) + (4 \times 0 \times 2) = 2 + 3 + 0 = 5$$

$$\det(A) = 8 - 5 = 3$$

Karena $\det(A) \neq 0$, matriks memiliki invers.

---

#### Langkah 2: Menghitung Matriks Kofaktor
Kofaktor diperoleh dengan menghitung nilai determinan minor $2\times2$ setelah menghapus baris ke-$i$ dan kolom ke-$j$, lalu dikalikan tanda selang-seling (+ / -).

$$\text{Kof}(A) = \begin{bmatrix}
+\begin{vmatrix} 1 & 1 \\ 3 & 4 \end{vmatrix} & -\begin{vmatrix} 0 & 1 \\ 2 & 4 \end{vmatrix} & +\begin{vmatrix} 0 & 1 \\ 2 & 3 \end{vmatrix} \\
-\begin{vmatrix} 2 & 1 \\ 3 & 4 \end{vmatrix} & +\begin{vmatrix} 1 & 1 \\ 2 & 4 \end{vmatrix} & -\begin{vmatrix} 1 & 2 \\ 2 & 3 \end{vmatrix} \\
+\begin{vmatrix} 2 & 1 \\ 1 & 1 \end{vmatrix} & -\begin{vmatrix} 1 & 1 \\ 0 & 1 \end{vmatrix} & +\begin{vmatrix} 1 & 2 \\ 0 & 1 \end{vmatrix}
\end{bmatrix}$$

Mari kita hitung satu per satu elemennya:
* $c_{11} = +(4 - 3) = 1$
* $c_{12} = -(0 - 2) = 2$
* $c_{13} = +(0 - 2) = -2$
* $c_{21} = -(8 - 3) = -5$
* $c_{22} = +(4 - 2) = 2$
* $c_{23} = -(3 - 4) = 1$
* $c_{31} = +(2 - 1) = 1$
* $c_{32} = -(1 - 0) = -1$
* $c_{33} = +(1 - 0) = 1$

Maka matriks kofaktornya adalah:
$$\text{Kof}(A) = \begin{bmatrix} 1 & 2 & -2 \\ -5 & 2 & 1 \\ 1 & -1 & 1 \end{bmatrix}$$

---

#### Langkah 3: Menentukan Matriks Adjoin ($\text{Adj}(A)$)
Adjoin adalah hasil transpose (tukar baris menjadi kolom) dari matriks kofaktor:
$$\text{Adj}(A) = (\text{Kof}(A))^T = \begin{bmatrix} 1 & -5 & 1 \\ 2 & 2 & -1 \\ -2 & 1 & 1 \end{bmatrix}$$

---

#### Langkah 4: Menghitung Invers Matriks $A$ ($A^{-1}$)
$$A^{-1} = \frac{1}{3} \begin{bmatrix} 1 & -5 & 1 \\ 2 & 2 & -1 \\ -2 & 1 & 1 \end{bmatrix}$$

$$A^{-1} = \begin{bmatrix} \frac{1}{3} & -\frac{5}{3} & \frac{1}{3} \\ \frac{2}{3} & \frac{2}{3} & -\frac{1}{3} \\ -\frac{2}{3} & \frac{1}{3} & \frac{1}{3} \end{bmatrix}$$

---

### Pembuktian ($A \times A^{-1} = I$)
$$\begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & 1 \\ 2 & 3 & 4 \end{bmatrix} \begin{bmatrix} \frac{1}{3} & -\frac{5}{3} & \frac{1}{3} \\ \frac{2}{3} & \frac{2}{3} & -\frac{1}{3} \\ -\frac{2}{3} & \frac{1}{3} & \frac{1}{3} \end{bmatrix} = \begin{bmatrix} \frac{1+4-2}{3} & \frac{-5+4+1}{3} & \frac{1-2+1}{3} \\ \frac{0+2-2}{3} & \frac{0+2+1}{3} & \frac{0-1+1}{3} \\ \frac{2+6-8}{3} & \frac{-10+6+4}{3} & \frac{2-3+4}{3} \end{bmatrix} = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{bmatrix}$$

Terbukti menghasilkan Matriks Identitas ($I$).

---

### Kesimpulan
Invers dari matriks $A$ adalah:
$$A^{-1} = \begin{bmatrix} \frac{1}{3} & -\frac{5}{3} & \frac{1}{3} \\ \frac{2}{3} & \frac{2}{3} & -\frac{1}{3} \\ -\frac{2}{3} & \frac{1}{3} & \frac{1}{3} \end{bmatrix}$$

## Soal 8
### Langkah Penyelesaian (Mencari Invers dengan Metode Adjoin)

Diketahui matriks:
$$B = \begin{bmatrix} 2 & 1 & 0 \\ 1 & 2 & 1 \\ 0 & 1 & 2 \end{bmatrix}$$

Rumus invers matriks adalah:
$$B^{-1} = \frac{1}{\det(B)} \cdot \text{Adj}(B)$$

---

#### Langkah 1: Menghitung Determinan Matriks $B$ ($\det(B)$)
Menggunakan Metode Sarrus, tulis kembali dua kolom pertama di sebelah kanan matriks:
$$\begin{matrix}
2 & 1 & 0 & | & 2 & 1 \\
1 & 2 & 1 & | & 1 & 2 \\
0 & 1 & 2 & | & 0 & 1
\end{matrix}$$

* **Total Diagonal Utama (\\):**
  $$(2 \times 2 \times 2) + (1 \times 1 \times 0) + (0 \times 1 \times 1) = 8 + 0 + 0 = 8$$
* **Total Diagonal Sekunder (/):**
  $$(0 \times 2 \times 0) + (1 \times 1 \times 2) + (2 \times 1 \times 1) = 0 + 2 + 2 = 4$$

$$\det(B) = 8 - 4 = 4$$

---

#### Langkah 2: Menghitung Matriks Kofaktor
Kita cari determinan minor $2\times2$ untuk setiap elemen dengan memberikan tanda selang-seling (+ / -):

## Soal 9
### Langkah Penyelesaian (Mencari Invers dengan Metode Adjoin)

Diketahui matriks:
$$C = \begin{bmatrix} 3 & 0 & 2 \\ 2 & 0 & -2 \\ 0 & 1 & 1 \end{bmatrix}$$

Rumus invers matriks adalah:
$$C^{-1} = \frac{1}{\det(C)} \cdot \text{Adj}(C)$$

---

#### Langkah 1: Menghitung Determinan Matriks $C$ ($\det(C)$)
Menggunakan Metode Sarrus, kita tulis kembali dua kolom pertama di sebelah kanan matriks:
$$\begin{matrix}
3 & 0 & 2 & | & 3 & 0 \\
2 & 0 & -2 & | & 2 & 0 \\
0 & 1 & 1 & | & 0 & 1
\end{matrix}$$

* **Total Diagonal Utama (\\):**
  $$(3 \times 0 \times 1) + (0 \times -2 \times 0) + (2 \times 2 \times 1) = 0 + 0 + 4 = 4$$
* **Total Diagonal Sekunder (/):**
  $$(0 \times 0 \times 2) + (1 \times -2 \times 3) + (1 \times 2 \times 0) = 0 + (-6) + 0 = -6$$

$$\det(C) = 4 - (-6) = 4 + 6 = 10$$

---

#### Langkah 2: Menghitung Matriks Kofaktor
Kita hitung nilai determinan minor $2\times2$ untuk setiap posisi dengan tanda selang-seling (+ / -):

$$\text{Kof}(C) = \begin{bmatrix}
+\begin{vmatrix} 0 & -2 \\ 1 & 1 \end{vmatrix} & -\begin{vmatrix} 2 & -2 \\ 0 & 1 \end{vmatrix} & +\begin{vmatrix} 2 & 0 \\ 0 & 1 \end{vmatrix} \\
-\begin{vmatrix} 0 & 2 \\ 1 & 1 \end{vmatrix} & +\begin{vmatrix} 3 & 2 \\ 0 & 1 \end{vmatrix} & -\begin{vmatrix} 3 & 0 \\ 0 & 1 \end{vmatrix} \\
+\begin{vmatrix} 0 & 2 \\ 0 & -2 \end{vmatrix} & -\begin{vmatrix} 3 & 2 \\ 2 & -2 \end{vmatrix} & +\begin{vmatrix} 3 & 0 \\ 2 & 0 \end{vmatrix}
\end{bmatrix}$$

Perhitungan masing-masing komponen:
* $c_{11} = +(0 - (-2)) = 2$
* $c_{12} = -(2 - 0) = -2$
* $c_{13} = +(2 - 0) = 2$
* $c_{21} = -(0 - 2) = 2$
* $c_{22} = +(3 - 0) = 3$
* $c_{23} = -(3 - 0) = -3$
* $c_{31} = +(0 - 0) = 0$
* $c_{32} = -(-6 - 4) = -(-10) = 10$
* $c_{33} = +(0 - 0) = 0$

Maka matriks kofaktornya adalah:
$$\text{Kof}(C) = \begin{bmatrix} 2 & -2 & 2 \\ 2 & 3 & -3 \\ 0 & 10 & 0 \end{bmatrix}$$

---

#### Langkah 3: Menentukan Matriks Adjoin ($\text{Adj}(C)$)
Adjoin diperoleh dengan melakukan transpose (menukar baris menjadi kolom) pada matriks kofaktor:
$$\text{Adj}(C) = (\text{Kof}(C))^T = \begin{bmatrix} 2 & 2 & 0 \\ -2 & 3 & 10 \\ 2 & -3 & 0 \end{bmatrix}$$

---

#### Langkah 4: Menghitung Invers Matriks $C$ ($C^{-1}$)
$$C^{-1} = \frac{1}{10} \begin{bmatrix} 2 & 2 & 0 \\ -2 & 3 & 10 \\ 2 & -3 & 0 \end{bmatrix}$$

$$C^{-1} = \begin{bmatrix} \frac{2}{10} & \frac{2}{10} & \frac{0}{10} \\ -\frac{2}{10} & \frac{3}{10} & \frac{10}{10} \\ \frac{2}{10} & -\frac{3}{10} & \frac{0}{10} \end{bmatrix} = \begin{bmatrix} \frac{1}{5} & \frac{1}{5} & 0 \\ -\frac{1}{5} & \frac{3}{10} & 1 \\ \frac{1}{5} & -\frac{3}{10} & 0 \end{bmatrix}$$

---

### Kesimpulan
Invers dari matriks $C$ adalah:
$$C^{-1} = \begin{bmatrix} \frac{1}{5} & \frac{1}{5} & 0 \\ -\frac{1}{5} & \frac{3}{10} & 1 \\ \frac{1}{5} & -\frac{3}{10} & 0 \end{bmatrix}$$
